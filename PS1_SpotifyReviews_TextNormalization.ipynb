```python
import pandas as pd
import nltk
import re
from nltk.corpus import brown
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
```

### Importing dataset

```python
pd.set_option('display.max_colwidth', None)
```

```python
# Import dataset
df = pd.read_csv('DATASET.csv', encoding="utf-8")

# Preview first 5 instances
df.head()
```

#### The Dataset

```python
df.info()
```

```python
df = df.drop(columns=['label'])
# Preview first 5 instances
df.head()
```

### Unitize and Tokenize

```python
# Handle missing values, apply word_tokenize to each entry in 'Review' column
df['tokenized_review'] = df['Review'].fillna('').apply(word_tokenize)

# Preview first 5 instances
df.head()
```

### Lowercase

```python
# Convert tokens to lowercase
df['tokenized_review'] = df['tokenized_review'].apply(lambda x: [token.lower() for token in x])

# Preview first 5 instances
df.head()
```

### Remove Tokens with Numbers

```python
df['tokenized_review'] = df['tokenized_review'].apply(lambda x: [token.lower() for token in x if not any(char.isdigit() for char in token)])

# Preview first 5 instances
df.head()
```

### Remove Special Characters and White Spaces

```python
import re

# Remove special characters from tokens
df['tokenized_review'] = df['tokenized_review'].apply(
    lambda x: [token for token in 
               (re.sub(r'[^a-zA-Z0-9]', '', t) for t in x)
               if token]
)

# Preview first 5 instances
df.head()
```

### Remove Stop Words

```python
# Download and import english stopwords
stop_words = stopwords.words('english')
print(stop_words)
```

```python
# Remove stopwords from tokens
df['tokenized_review'] = df['tokenized_review'].apply(
    lambda x: [token for token in x if token.lower() not in stop_words]
)

# Preview first 5 instances
df.head()
```

### Stemming

```python
from nltk.stem import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# Apply stemming to tokens
df['stemmed_review'] = df['tokenized_review'].apply(lambda x: [stemmer.stem(token) for token in x])

# Preview first 5 instances
df.head()
```

### Lemmatization

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')

# Initialize WordNet Lemmatizer
lemmatizer = WordNetLemmatizer()

# Apply lemmatization to each token
df['lemmatized_review'] = df['tokenized_review'].apply(lambda x: [lemmatizer.lemmatize(token) for token in x])

# Preview first 3 instances
df.head()
```
